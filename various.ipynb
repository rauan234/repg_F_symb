```python
data = read_from_file("D8_data.json")
```

```python
# load untwisted rep data for G
repg = load_repg(data["G"])

# compute Rep(G) F-symbols and check unitarity and pentagon equation
repg.compute_F_symb()
repg.check_F_unitary()
repg.check_pentagon_eqs()
```

```python
# pick a subgroup and compute its irreps for each cohomology class
subg_n = 0  # choose the 0th subgroup
print("Data for this subgroup:", data["subgroups"][subg_n])
reph_list = load_subgroup(repg, data["subgroups"][subg_n])
print(f"n cohomology classses = {len(reph_list)}")

coh_n = 0  # pick the 0th cohomology class
reph = reph_list[coh_n]
print(reph)
```

```python
# compute module F-symbols and verify unitarity and pentagon equations
reph.compute_F_right()
reph.compute_F_left()
reph.compute_F_lr()

reph.check_F_unitary()
reph.check_pentagon_eqs()
```
